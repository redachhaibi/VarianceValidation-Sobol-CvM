```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from sympy import latex
```

```python
#Function that compute the distribution function at x of a Gaussian random variable with mean m and variance sig2
def gaussian_distrib(x,m,sig2):
    t=sp.symbols('t')
    return sp.integrate(sp.exp(-(t-m)*(t-m)/(2*sig2))/(sp.sqrt(2*sp.pi*sig2)), (t, -sp.oo, x))
```

```python
#Function that integrate a symbolic function ff on the variable x on R using the standard Gaussian measaure
def esperance(ff, x):
  
#    ff : expression sympy
#        Fonction à intégrer
#    x : symbol sympy        

    return sp.integrate(ff*sp.exp(-x*x/2)/(sp.sqrt(2*sp.pi)), (x, -sp.oo, sp.oo))
```

```python
# Symbolic variables definition
x1, x2 = sp.symbols('x1 x2',real=True)
ro = sp.symbols('ro',real=True)
sig_residuel2=sp.symbols('sig_residuel',real=True,positive=True)
#sig_residuel2=1-ro*ro
```

```python
#Computation of the Cramer Von Mises index for Y=ro*X1+sqrt(1-ro^2)*X2 where X1 and X2 are two independent standard normal
y=sp.symbols('y',real=True)
integrand=(gaussian_distrib(y,0,1)-gaussian_distrib(y,ro*x1,sig_residuel2))
integrand=sp.simplify(integrand)
integrand=sp.factor(integrand)
integrand=integrand**2;
first_integral=esperance(integrand,x1)
#print(latex(integrand))
```
