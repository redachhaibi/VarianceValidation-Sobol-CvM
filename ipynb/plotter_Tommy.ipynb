```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Lectures des données
alpha_vals=np.loadtxt("alpha_vals_varestim.dat");
cible_tab=np.loadtxt("cible_CVM.dat");
var_est=np.loadtxt("var_est_CVM.dat");
Chatestim=np.loadtxt("Chatestim.dat");
var_theo=np.loadtxt("var_theo_CVM.dat");
```

```python
# Create two plot asided
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# ----- first plot -----
axes[0].plot(alpha_vals,cible_tab,alpha_vals,Chatestim)
axes[0].set_title("CVM index in the Tommy Model")
axes[0].set_xlabel(r"$\alpha$")
axes[0].set_ylabel("CVM index and its Chatterjee estimate")
axes[0].grid(True)

# ----- second plot -----
axes[1].plot(alpha_vals,var_theo,alpha_vals,var_est)
axes[1].set_title("Variance of the Chatterjee estimator in the Tommy Model")
axes[1].set_xlabel(r"$\alpha$")
axes[1].set_ylabel("Asymptotic variance and its estimate")
axes[1].grid(True)

# Ajusting spaces automatically
plt.tight_layout()

#saving the plot
plt.savefig("graph_CVM_Tommy.png", dpi=300)
#Display
plt.show()
```
