```python
import sympy as sp
from sympy import latex
# Déclaration des variables symboliques
x1, x2, x3 = sp.symbols('x1 x2 x3')
a, b = sp.symbols('a b')
```

```python
# Définition de la fonction d'Ishigami
f = sp.sin(x1) + a * sp.sin(x2)**2 + b * x3**4 * sp.sin(x1)
#f = a*sp.sin(x1) +  sp.sin(x2)**2 + b * sp.sin(x3) * sp.sin(x1)
```

```python
# Calcul de l'intégrale de f sur x2 et x3 dans [-pi, pi]
integralef_x3 = sp.integrate(f,(x3, -sp.pi, sp.pi))
integralef_x2_x3 = sp.integrate(integralef_x3,(x2, -sp.pi, sp.pi))
#normalisation pour l'espérance conditionnelle de 
esperance_condf = integralef_x2_x3 / (2*sp.pi)**2
# Calcul de l'intégrale de f² sur x2 et x3 dans [-pi, pi]
f2=f*f
integralef2_x3 = sp.integrate(f2,(x3, -sp.pi, sp.pi))
integralef2_x2_x3 = sp.integrate(integralef2_x3,(x2, -sp.pi, sp.pi))
esperance_condf2 = integralef2_x2_x3 / (2*sp.pi)**2
f3=f2*f
integralef3_x3 = sp.integrate(f3,(x3, -sp.pi, sp.pi))
integralef3_x2_x3 = sp.integrate(integralef3_x3,(x2, -sp.pi, sp.pi))
esperance_condf3 = integralef3_x2_x3 / (2*sp.pi)**2
f4=f3*f
integralef4_x3 = sp.integrate(f4,(x3, -sp.pi, sp.pi))
integralef4_x2_x3 = sp.integrate(integralef4_x3,(x2, -sp.pi, sp.pi))
esperance_condf4 = integralef4_x2_x3 / (2*sp.pi)**2
sigma2=esperance_condf2-esperance_condf*esperance_condf
```

```python
#Calcul de mB
mB1=sp.integrate(esperance_condf*esperance_condf,(x1, -sp.pi, sp.pi))/ (2*sp.pi);
mB2=sp.integrate(esperance_condf,(x1, -sp.pi, sp.pi))/ (2*sp.pi);
mB3=sp.integrate(esperance_condf2,(x1, -sp.pi, sp.pi))/ (2*sp.pi);
mB= sp.Matrix([mB1, mB2, mB3])
```

```python
#Calcul de SigmaB
SigmaB11=sp.integrate(sigma2*sigma2,(x1, -sp.pi, sp.pi))/ (2*sp.pi)+4*sp.integrate(sigma2*esperance_condf*esperance_condf,(x1, -sp.pi, sp.pi))/ (2*sp.pi)
SigmaB22=sp.integrate(sigma2,(x1, -sp.pi, sp.pi))/ (2*sp.pi)
SigmaB33=sp.integrate(esperance_condf4,(x1, -sp.pi, sp.pi))/ (2*sp.pi)-(sp.integrate(esperance_condf2,(x1, -sp.pi, sp.pi))/ (2*sp.pi))**2
SigmaB21=2*sp.integrate(sigma2*esperance_condf,(x1, -sp.pi, sp.pi))/ (2*sp.pi)
SigmaB12=SigmaB21
SigmaB13=2*sp.integrate(esperance_condf3*esperance_condf,(x1, -sp.pi, sp.pi))/ (2*sp.pi)-2*sp.integrate(esperance_condf2*esperance_condf*esperance_condf,(x1, -sp.pi, sp.pi))/ (2*sp.pi)
SigmaB31=SigmaB13
SigmaB23=sp.integrate(esperance_condf3,(x1, -sp.pi, sp.pi))/ (2*sp.pi)-sp.integrate(esperance_condf2*esperance_condf,(x1, -sp.pi, sp.pi))/ (2*sp.pi)
SigmaB32=SigmaB23
SigmaB= sp.Matrix([[SigmaB11,SigmaB12,SigmaB13], [SigmaB21,SigmaB22,SigmaB23], [SigmaB31,SigmaB32,SigmaB33]]);
```

```python
#Calcul de SigmaC
SigmaC11=sp.integrate(esperance_condf**4,(x1, -sp.pi, sp.pi))/ (2*sp.pi)-mB1*mB1
SigmaC22=mB1-mB2*mB2
SigmaC33=SigmaC11
SigmaC21=sp.integrate(esperance_condf**3,(x1, -sp.pi, sp.pi))/ (2*sp.pi)-mB1*mB2
SigmaC12=SigmaC21
SigmaC13=SigmaC11
SigmaC31=SigmaC13
SigmaC23=SigmaC12
SigmaC32=SigmaC23
SigmaC= sp.Matrix([[SigmaC11,SigmaC12,SigmaC13], [SigmaC21,SigmaC22,SigmaC23], [SigmaC31,SigmaC32,SigmaC33]]);
```

```python
#calcul de g
esperance=mB2;
variance=sp.integrate(esperance_condf2,(x1, -sp.pi, sp.pi))/ (2*sp.pi)-esperance*esperance
zobol=(mB1-mB2*mB2)/variance
g=sp.Matrix([1,2*esperance*(zobol-1),-zobol])/variance
```

```python
#calcul de la variance asymptotique (Bernouillade)
sigma_asymp=g.T*(SigmaB+SigmaC)*g
```

```python
#En ne gardant que les premières lignes et premières colonnes de SigmaB et SigmaC et on remplace le reste par les éléments
#classiques pour la moyenne empirique et la moyenne des carrés
masque=sp.Matrix([[1,1,1], [1,0,0], [1,0,0]])
Macovariance=sp.hadamard_product(SigmaB+SigmaC,masque)
Macovariance22=variance
Macovariance33=sp.integrate(esperance_condf4,(x1, -sp.pi, sp.pi))/ (2*sp.pi)-mB3*mB3
Macovariance23=sp.integrate(esperance_condf3,(x1, -sp.pi, sp.pi))/ (2*sp.pi)-mB1*mB3
Macovariance=Macovariance+sp.Matrix([[0,0,0], [0,Macovariance22,Macovariance23], [0,Macovariance23,Macovariance33]])
Mon_sigma_asymp=g.T*(Macovariance)*g
```

```python
#évalution de l'indice de Sobol' et de la variance asymptotique dans le cas a=10, b=0.1
#avec les calculs de l'article Bernoulli et son errata
values = {a: 10., b: 1}
Zobol_num=zobol.evalf(subs=values);
sigma_Zobol_hat_num=sigma_asymp.evalf(subs=values);
Mon_sigma_Zobol_hat_num=Mon_sigma_asymp.evalf(subs=values);
print('Dans le cas a=10. et b=1');
print('indice de Sobol vaut: ', Zobol_num);
print('variance asymptotique Bernoulli vaut: ',sigma_Zobol_hat_num)
print('Ma variance asymptotique vaut: ',Mon_sigma_Zobol_hat_num)
```
