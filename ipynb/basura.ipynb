```python
def sigma(alpha):


    # Première intégrale
    Ixx = sp.integrate(
        sp.integrate(
            C_XX(t, s, alpha) * f_Y(t, alpha) * f_Y(s, alpha),
            (t, 0, 1)
        ),
        (s, 0, 1)
    )

    # Deuxième intégrale
    Iyy = sp.integrate(
        sp.integrate(
            C_YY(t, s, alpha) * f_T(t, alpha) * f_T(s, alpha),
            (t, 0, 1)
        ),
        (s, 0, 1)
    )

    # Troisième intégrale (découpage t<s et s<=t)
    Ixy = (
        sp.integrate(
            sp.integrate(
                C_XY(t, s, alpha) * f_Y(t, alpha) * f_T(s, alpha),
                (t, 0, s)
            ),
            (s, 0, 1)
        )
        +
        sp.integrate(
            sp.integrate(
                C_XY(t, s, alpha) * f_T(t, alpha) * f_S(s, alpha),
                (t, s, 1)
            ),
            (s, 0, 1)
        )
    )

    return sp.simplify(36 * (Ixx + Iyy - 2*Ixy))
```
