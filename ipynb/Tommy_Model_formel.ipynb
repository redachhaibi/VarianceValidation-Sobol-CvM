```python
import sympy as sp

# Variables symboliques
alpha, t, s = sp.symbols('alpha t s', real=True)
```

```python
# ------------------------------------------------------------------
# C_{X,X}(t,s)
# ------------------------------------------------------------------

def C_XX(t, s, alpha):


    u = sp.Min(s, t)
    v = sp.Max(s, t)

    return (
        (2*alpha*u**2*v)/((alpha-4)*(alpha-3))
        * (alpha-4 - (alpha-3)*v + v**(alpha-3))
        + (u/(alpha-2))
        * (alpha*u - 2*u**(alpha-1))
        * (
            1
            - (v/(alpha-2))
            * (alpha*v - 2*v**(alpha-1))
        )
    )
```

```python
# ------------------------------------------------------------------
# C_{Y,Y}(t,s)
# ------------------------------------------------------------------

def C_YY(t, s, alpha):


    u = sp.Min(s, t)
    v = sp.Max(s, t)

    return (
        (alpha*u - u**alpha)
        * (alpha - 1 - alpha*v + v**alpha)
        / (alpha - 1)**2
    )
```

```python
# ------------------------------------------------------------------
# C_{X,Y}(t,s)
# ------------------------------------------------------------------

def C_XY(t, s, alpha):


    # Cas s <= t
    expr_st = (
        alpha * (
            2*s**alpha/(alpha*(alpha-2))
            - s**2/(alpha-3)
            * (
                t
                - t**(alpha-2)/(alpha-2)
            )
        )
        + (s/(alpha-2))
        * (alpha*s - 2*s**(alpha-1))
        * (
            2
            - (alpha*t - t**alpha)/(alpha-1)
        )
    )

    # Cas t < s
    expr_ts = (
        alpha*t * (
            t**(alpha-1)/(alpha*(alpha-1))
            + s**2/(alpha-3)
            * (
                2*s**(alpha-3)/(alpha-1)
                - 1
            )
        )
        + s/((alpha-2)*(alpha-1))
        * (alpha*s - 2*s**(alpha-1))
        * (t**alpha - alpha*t)
        + (2*t)/((alpha-1)*(alpha-2))
        * (
            alpha*(alpha-1)*s
            - (
                alpha*s**(alpha-1)
                + (alpha-2)*t**(alpha-1)
            )
        )
    )

    return sp.Piecewise(
        (expr_st, s <= t),   # inclut le cas s=t
        (expr_ts, True)
    )
```

```python
# ------------------------------------------------------------------
# Densité f_Y(s)
# ------------------------------------------------------------------

def f_Y(s, alpha):


    return alpha/(alpha - 1) * (1 - s**(alpha - 1))
```

```python
# ------------------------------------------------------------------
# Densité f_T(t)
# ------------------------------------------------------------------

def f_T(t, alpha):


    return (2*alpha*(t - t**(alpha - 1))) / (alpha - 2)
```
