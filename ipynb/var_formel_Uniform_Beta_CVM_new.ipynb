```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from sympy import latex
```

```python
#defining the  density  of (X,Y) depending on a>0
#u is a real input
#the output is the density of (X,Y) at u,v
def densXY(u, v, aa):
    dens = (
        sp.Piecewise(
            ((aa)/(2)*u**(aa-2), (u >= 0) & (u <= 1)),
            (0, True)
        )
        *
        sp.Piecewise(
            (1, (v > -u) & (v < u)),
            (0, True)
        )
    )
    
    return dens
```

```python
# Symbolic variables definition
x,y,t,s= sp.symbols('x y t s')
#
a = sp.symbols('a', positive=True, real=True)
```

```python
#Calcul des différentes densités 
dXY=densXY(x,y,a);
dX=sp.integrate(dXY,(y,0,1));
dY=sp.integrate(dXY,(x,0,1));
dY_sach_X=dXY/dX;
```

```python
#Calcul des fonctions de répartition utiles et de la mesure R
FX=sp.integrate(dX,(x,0,t));
FYt=sp.integrate(dY,(y,-1,t));
FYs=sp.integrate(dY,(y,-1,s));
FY_sach_X=sp.integrate(dY_sach_X,(y,-1,t));
temp=(FY_sach_X**2)*dX;
temp=sp.simplify(temp)
RYt=sp.integrate(temp,(x,0,1));
RYt=sp.simplify(RYt);
dRYt=sp.diff(RYt, t);
dRYt=sp.simplify(dRYt);
RYs = RYt.subs(t, s);
dRYs=dRYt.subs(t, s);
```

```python
#dessiner la fonctionn de répartition de Y
FYt_num = FYt.subs(a, 2)
# Conversion en fonction numpy
FYt_num = sp.piecewise_fold(FYt_num)
RepY = sp.lambdify(t, FYt_num, "numpy")
t_vals= np.arange(-2, 2.01, 0.01);
ff = np.vectorize(RepY)(t_vals)
plt.plot(t_vals,ff)
```

```python
#dessiner la fonctionn de répartition de X
FX_num = FX.subs(a, 1)
# Conversion en fonction numpy
FX_num = sp.piecewise_fold(FX_num)
RepX = sp.lambdify(t, FX_num, "numpy")
t_vals= np.arange(-1, 2.01, 0.01);
ff = np.vectorize(RepX)(t_vals)
plt.plot(t_vals,ff)
```

```python
#dessiner la densité jointe


densXY_num = dXY.subs(a, 4)

# Conversion en fonction numpy
fXY = sp.lambdify((x, y), densXY_num, "numpy")

# Grille
X = np.linspace(0, 1, 300)
Y = np.linspace(-1, 1, 300)

XX, YY = np.meshgrid(X, Y)

# Valeurs de la densité
ZZ = fXY(XX, YY)

# Figure 3D
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(XX, YY, ZZ)

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("densité jointe")

plt.show()
```
