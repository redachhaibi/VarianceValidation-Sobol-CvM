```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Initialisation des constantes
#
Ntrysample=3000;
Nhorizon=20000;
Nhorizonm1=Nhorizon-1;
Nnormalize=3/(Nhorizon*Nhorizon);
```

```python
def Estim_var(aalph,cible):
    SS1=np.zeros((Ntrysample));
#
    for i in range(Ntrysample): 
        AA=[];
        rangos=[];
        FAA=[];
        FAA1=[];
        FAA1rangos=[];
        AA=np.random.beta(aalph,1,Nhorizon);
        UU=np.random.uniform(0,1,Nhorizon);
        FAA=AA*UU;
#enumére les adresses plus petit au plus grand
        rangos=np.argsort(AA);
#On réarrange suivant l'ordre des échantillons pour chacune des variables    
        FAA1rangos=FAA[rangos];
#Calcul de l'estimateur CVM de Chatterjee 
        rangoYordreX=np.argsort(np.argsort(FAA1rangos))
        Shift_rangoYordreX=np.zeros(Nhorizon)
        Shift_rangoYordreX[0:Nhorizonm1]=rangoYordreX[1:Nhorizon];
        tempestim=1-Nnormalize*np.sum(np.absolute(rangoYordreX-Shift_rangoYordreX))
        SS1[i]=tempestim-cible;
    v=Nhorizon*np.var(SS1)
    return v,tempestim
```

```python
alpha_vals = np.arange(0.5, 7., 0.07);
Nbmax=alpha_vals.shape[0];
Chatestim=np.zeros(Nbmax);
var_est=np.zeros(Nbmax);
cible_tab=np.zeros(Nbmax);
i=0;
for aa in alpha_vals:
    c=2/((aa+1)*(aa+2));
    cible_tab[i]=c;
    var_est[i],Chatestim[i]=Estim_var(aa,c);
    i=i+1;
```

```python
#plt.plot(alpha_vals,cible_tab, alpha_vals,Chatestim);
plt.plot(alpha_vals,var_est)
```
