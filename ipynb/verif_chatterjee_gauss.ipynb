```python
import numpy as np
import numpy.random as npr
import scipy.stats as scista
import matplotlib.pyplot as plt
```

```python
# N1 est la taille de l'échantillon Monte Carlo général-N2 la taille de l'échantillon Monte Carlo pour la loi conditionnelle 
# NN est le nombre de point pour calculer en méthode de Riemann l'écart entre la FR gaussienne et la FR conditionnelle
N1=5000;
N2=5000;
NN=1000;
ro=1;
ro1=np.sqrt(1-ro*ro);
ro2=1+ro*ro
truecvm=1-(6/np.pi)*np.arcsin(ro1/2);
truecvmbis=-1/2+(3/np.pi)*np.arcsin(ro2/2);
#
DeltaX=np.zeros(N1);
epsilon1=npr.normal(0,1,N1);
epsilon2=npr.normal(0,1,N1);
#Simulation de X
X=epsilon1;
#Simulation de E(Y|X)
ychap=ro*X;
#Simulation de Y
Y=ychap+ro1*epsilon2;
#Calcul de la fonction de répartition empirique de Y
FGauss=scista.ecdf(Y);
#Discrétisation de l'intervalle [-5,5]
t=np.linspace(-5.0, 5, num=NN);
pa=10/NN;
#Calcul de la densité gaussienne standard sur les points de discrétisation
rt2pi=np.sqrt(2*np.pi);
gt=(np.exp(-0.5*t*t))/rt2pi;
#Calcul de l'estimateur de la fonction de répartition sur les points de discrétisation
FGaussechan=FGauss.cdf.evaluate(t)
#Boucle de génération des réalisations conditionnelles à chaque ycond on calcul l'écart L2(N(0,1)) entre 
# la fonction de répartition conditionnelle et la non conditionnelle . Cet écart est calculer par une approximation de Riemann
for i in range(1,N1):
    m=ychap[i];
#génération d'un échantillon conditionnel    
    epsilon3=npr.normal(0,1,N2);
    Ycond=m+ro1*epsilon3;
#Calcul de la fonction de répartition conditionnelle et de son écart quadratique intégré
    FGausscond=scista.ecdf(Ycond);
    Delta=(FGausscond.cdf.evaluate(t)-FGaussechan);
    Delta=Delta*Delta*gt
    DeltaX[i]=np.sum(Delta)*pa;
#Intégration Monte Carlo en X
CvM=np.mean(DeltaX);  
print(CvM*6)
print(truecvm)
print(truecvmbis)
```

```python
ro_vals = np.arange(-1, 1.01, 0.01);
ro2_vals=(1+ro_vals*ro_vals)/2;
truecvm=-1/2+(3/np.pi)*np.arcsin(np.minimum(ro2_vals,1));
#print(ro_vals)
#print(truecvm)
```

```python
plt.plot(ro_vals,truecvm,ro_vals,ro_vals**2)
plt.title("CVM index (blue) and Sobol' index (orange) in the Gaussian Model")
plt.xlabel(r"$\rho$")
plt.ylabel("CVM and Sobol'")
plt.grid(True)
plt.savefig("graph_CVM_Gaussian.png", dpi=300)
plt.show()
```
