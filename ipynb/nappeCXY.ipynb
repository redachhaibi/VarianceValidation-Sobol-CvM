```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# ------------------------------------------------------------
# alpha fixé
# ------------------------------------------------------------
alpha = 50

# ------------------------------------------------------------
# définition de C_{X,Y}
# ------------------------------------------------------------
def C_XY_numeric(t, s):

    if s <= t:
        return (
            alpha * (
                2*s**alpha/(alpha*(alpha-2))
                - s**2/(alpha-3)
                * (t - t**(alpha-2)/(alpha-2))
            )
            + s/(alpha-2)
            * (alpha*s - 2*s**(alpha-1))
            * (2 - (alpha*t - t**alpha)/(alpha-1))
        )

    else:
        return (
            alpha*t * (
                t**(alpha-1)/(alpha*(alpha-1))
                + s**2/(alpha-3)
                * (2*s**(alpha-3)/(alpha-1) - 1)
            )
            + s/((alpha-2)*(alpha-1))
            * (alpha*s - 2*s**(alpha-1))
            * (t**alpha - alpha*t)
            + (2*t)/((alpha-1)*(alpha-2))
            * (
                alpha*(alpha-1)*s
                - (alpha*s**(alpha-1)
                   + (alpha-2)*t**(alpha-1))
            )
        )

# ------------------------------------------------------------
# grille sur [0,1]^2
# ------------------------------------------------------------
n = 120
t_vals = np.linspace(0, 1, n)
s_vals = np.linspace(0, 1, n)

T, S = np.meshgrid(t_vals, s_vals)

Z = np.zeros_like(T)

# ------------------------------------------------------------
# calcul de la surface
# ------------------------------------------------------------
for i in range(n):
    for j in range(n):
        Z[i, j] = C_XY_numeric(T[i, j], S[i, j])

# ------------------------------------------------------------
# plot 3D
# ------------------------------------------------------------
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(T, S, Z, cmap='viridis', edgecolor='none')

ax.set_title(r"Surface de $C_{X,Y}(t,s)$ pour $\alpha$=50")
ax.set_xlabel("t")
ax.set_ylabel("s")
ax.set_zlabel(r"$C_{X,Y}(t,s)$")

plt.show()
```
